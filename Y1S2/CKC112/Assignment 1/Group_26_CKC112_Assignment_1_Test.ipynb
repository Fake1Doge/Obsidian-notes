#C++ Code\n

````python
%%writefile Person.h\n#ifndef PERSON_H\n#define PERSON_H\n\n#include <string>\n\nclass Person {\nprotected:\n    std::string id;\n    std::string name;\n    std::string state;\n    std::string email;\n\npublic:\n    Person() : id(""), name(""), state(""), email("") {}\n    Person(std::string i, std::string n, std::string s, std::string e) \n        : id(i), name(n), state(s), email(e) {}\n\n    std::string getId() const { return id; }\n    std::string getName() const { return name; }\n};\n\n#endif\n
````

````python
%%writefile Review.h\n#ifndef REVIEW_H\n#define REVIEW_H\n\n#include <iostream>\n#include <string>\n\nclass Review {\nprivate:\n    std::string userId;\n    int rating;\n    std::string reviewText;\n    std::string hotelName;\n\npublic:\n    Review() : userId(""), rating(0), reviewText(""), hotelName("") {}\n    Review(std::string id, int r, std::string text, std::string hName) \n        : userId(id), rating(r), reviewText(text), hotelName(hName) {}\n\n    std::string getUserId() const { return userId; }\n    int getRating() const { return rating; }\n    std::string getReviewText() const { return reviewText; }\n    std::string getHotelName() const { return hotelName; }\n\n    void display() const {\n        std::cout << "Hotel: " << hotelName << " | Rating: " << rating << "/5\n";\n        std::cout << "Review: " << reviewText << "\n";\n    }\n\n    // Overloaded > operator for rating comparisons\n    bool operator>(const Review& other) const {\n        return this->rating > other.rating;\n    }\n};\n\n#endif\n
````

````python
%%writefile Traveler.h\n#ifndef TRAVELER_H\n#define TRAVELER_H\n\n#include "Person.h"\n#include "Review.h"\n#include <iostream>\n\nclass Traveler : public Person {\nprivate:\n    std::string memberType;\n    int points;\n    Review reviews[50]; // Fixed-size array (Has-A Composition)\n    int reviewCount;\n\npublic:\n    Traveler() : Person(), memberType(""), points(0), reviewCount(0) {}\n    Traveler(std::string id, std::string name, std::string state, std::string email, std::string mType, int p)\n        : Person(id, name, state, email), memberType(mType), points(p), reviewCount(0) {}\n\n    void addReview(const Review& r) {\n        if (reviewCount < 50) {\n            reviews[reviewCount++] = r;\n        }\n    }\n\n    void displayProfile() const {\n        std::cout << "--- Traveler Profile ---\n";\n        std::cout << "Name: " << name << " (" << id << ")\n";\n        std::cout << "Type: " << memberType << " | Points: " << points << "\n";\n        std::cout << "Authored Reviews: " << reviewCount << "\n";\n        for (int i = 0; i < reviewCount; i++) {\n            std::cout << "- ";\n            reviews[i].display();\n        }\n        std::cout << "------------------------\n";\n    }\n\n    int getReviewCount() const { return reviewCount; }\n    const Review& getReview(int index) const { return reviews[index]; }\n\n    // Friend Function Declaration\n    friend void generateTravelerReport(const Traveler& t);\n};\n\n// Friend Function Implementation\nvoid generateTravelerReport(const Traveler& t) {\n    std::cout << "Report: Traveler " << t.name << " currently holds " << t.points << " loyalty points.\n";\n}\n\n#endif\n
````

````python
%%writefile main.cpp\n#include <iostream>\n#include <fstream>\n#include <sstream>\n#include <string>\n#include "Traveler.h"\n\nusing namespace std;\n\n// Function Prototypes\nvoid propertyDeepDive(const Traveler travelers[], int travelerCount);\nvoid travelerProfile(Traveler travelers[], int travelerCount);\nvoid topPickMatcher(const Traveler travelers[], int travelerCount);\n\nint main() {\n    Traveler travelers[100]; // Single primary entity object (Fixed-Size Array)\n    int travelerCount = 0;\n\n    // One-Time I/O: Read users.txt\n    ifstream userFile("users.txt");\n    if (userFile.is_open()) {\n        string line;\n        while (getline(userFile, line)) {\n            stringstream ss(line);\n            string id, name, state, email, mType;\n            int points;\n\n            getline(ss, id, '\t');\n            getline(ss, name, '\t');\n            getline(ss, state, '\t');\n            getline(ss, email, '\t');\n            getline(ss, mType, '\t');\n            ss >> points;\n            \n            travelers[travelerCount++] = Traveler(id, name, state, email, mType, points);\n        }\n        userFile.close();\n    }\n\n    // One-Time I/O: Read reviews.txt\n    ifstream reviewFile("reviews.txt");\n    if (reviewFile.is_open()) {\n        string line;\n        while (getline(reviewFile, line)) {\n            stringstream ss(line);\n            string uId, reviewText, hotelName;\n            int rating;\n            \n            getline(ss, uId, '\t');\n            ss >> rating;\n            ss.ignore(); // ignore tab\n            getline(ss, reviewText, '\t');\n            getline(ss, hotelName, '\t');\n\n            Review r(uId, rating, reviewText, hotelName);\n            for (int i = 0; i < travelerCount; i++) {\n                if (travelers[i].getId() == uId) {\n                    travelers[i].addReview(r);\n                    break;\n                }\n            }\n        }\n        reviewFile.close();\n    }\n\n    int choice;\n    do {\n        cout << "\n=== Hotel Review System Menu ===\n";\n        cout << "1. Property Deep-Dive\n";\n        cout << "2. Traveler Profile\n";\n        cout << "3. Top-Pick Matcher\n";\n        cout << "4. Exit\n";\n        cout << "Enter choice: ";\n        cin >> choice;\n\n        switch (choice) {\n            case 1: propertyDeepDive(travelers, travelerCount); break;\n            case 2: travelerProfile(travelers, travelerCount); break;\n            case 3: topPickMatcher(travelers, travelerCount); break;\n            case 4: cout << "Exiting system...\n"; break;\n            default: cout << "Invalid choice.\n";\n        }\n    } while (choice != 4);\n\n    // One-Time Output I/O\n    ofstream outFile("system_log.txt");\n    if (outFile.is_open()) {\n        outFile << "System execution completed successfully. " << travelerCount << " travelers were loaded into memory.\n";\n        outFile.close();\n    }\n\n    return 0;\n}\n\nvoid propertyDeepDive(const Traveler travelers[], int travelerCount) {\n    string hotel;\n    cout << "Enter exact hotel name (e.g., Grand Imperial Hotel): ";\n    cin.ignore();\n    getline(cin, hotel);\n\n    int totalReviews = 0;\n    int sumRating = 0;\n\n    cout << "\n--- Reviews for " << hotel << " ---\n";\n    for (int i = 0; i < travelerCount; i++) {\n        for (int j = 0; j < travelers[i].getReviewCount(); j++) {\n            Review r = travelers[i].getReview(j);\n            if (r.getHotelName() == hotel) {\n                r.display();\n                sumRating += r.getRating();\n                totalReviews++;\n            }\n        }\n    }\n    \n    if (totalReviews > 0) {\n        cout << "\n=> Average Rating: " << (double)sumRating / totalReviews << "/5\n";\n    } else {\n        cout << "No reviews found for this hotel.\n";\n    }\n}\n\nvoid travelerProfile(Traveler travelers[], int travelerCount) {\n    string uId;\n    cout << "Enter Traveler ID (e.g., U001): ";\n    cin >> uId;\n\n    for (int i = 0; i < travelerCount; i++) {\n        if (travelers[i].getId() == uId) {\n            travelers[i].displayProfile();\n            generateTravelerReport(travelers[i]); // Friend Function call\n            return;\n        }\n    }\n    cout << "Traveler not found.\n";\n}\n\nvoid topPickMatcher(const Traveler travelers[], int travelerCount) {\n    string bestHotel = "";\n    int bestRating = -1;\n\n    for (int i = 0; i < travelerCount; i++) {\n        for (int j = 0; j < travelers[i].getReviewCount(); j++) {\n            Review r = travelers[i].getReview(j);\n            Review temp("", bestRating, "", "");\n            \n            // Demonstrate operator overloading '>'\n            if (r > temp) {\n                bestRating = r.getRating();\n                bestHotel = r.getHotelName();\n            }\n        }\n    }\n\n    if (bestRating != -1) {\n        cout << "Top Rated Hotel Found (By Highest Single Review): " << bestHotel << " (" << bestRating << "/5)\n";\n    } else {\n        cout << "No reviews available to match.\n";\n    }\n}\n
````
