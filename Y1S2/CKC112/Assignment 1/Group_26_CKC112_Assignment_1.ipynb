#Project Title: Hotel Review System



##1.0 Project Specification
---
The Hotel Review System is a console-based application designed to stimulate a real-world platform for managing hotel reviews and traveler informations. Its main objective is to efficiently store, organize and analyze review data so that users can gain insights into hotel performance and traveler behaviour. The system functions as an administrative tool with a menu-driven interface, without requiring login or multi-user management.

The system uses two input files including Users.txt for traveler informations and Reviews.txt for hotel reviews. These files are read once at the start and stored in memory using object arrays. All operations are performed in memory and data is written back to files once before the program ends following the One-Time I/O Rule.

Key features include Property Deep-Dive for viewing hotel reviews and average ratings, Traveler Profile for displaying user details and contributions, Top-Pick Matcher for identifying and ranking the best hotels, and Top Reviewers for analyzing user activity. The system applies Object Oriented Programming concepts such as inheritence, composition, operator overloading and friend function while following constraints like using fixed-size arrays and avoiding STL and polymorphism.


##2.0 Problem Analysis
---
The main problem addressed is the difficulty of managing and analyzing large amounts of hotel review data. Without a structured system, it is hard to extract useful insights such as identifying top hotels or understanding traveler contributions.

To solve this, the program reads and stores data from files into object arrays, ensuring organized data management. It supports searching and filtering through the Property Deep-Dive feature which allowing users to view reviews and calculate average ratings. The Traveler Profile feature helps track user information and contributions while the Top-Pick Matcher ranks hotel based on ratings. The Top Reviewers feature identifies the most active users.

Object Oriented Programming is used to structure the solution effectively. Inheritance extends traveler functionality, composition models relationships between travelers and reviews, operator overloading simplifies comparisons and friend functions enable reporting access. The system also follows all technical constraints, ensuring efficient proccessing and a clean design.

Overall, the program transform raw review data into meaningful insights through structured design console and Object Oriented Programming principles.

## 3.0 Constraints and Assumptions
---
1. Constraints
    * **Array Size Limits** - This program relies on fixed-size static array.
        * Maximum Users: 10(No new users will be added in)
        * Maximum Reviews: 500(Each traveller can only make maximum 50 reviews)
    * **File Format** - All text files must strictly use **Tab(\t)** as separator. Using spaces or other separator will cause read errors.

2. Assumptions
    *   **Unique ID** - Assume that every Traveller have an unique User ID in ***users.txt***.
    *   **Valid File Structure** - Assume that the text file exist in the same directory and follow the correct column order for the data.
    *   **English Language** - Assume all the inputs and file content are in English character(ASCII).



##4.0 Programming Design & Solution

###4.1 Process Identification
---
1. System Initialization
   - Program boots and allocates fixed-size arrays in memory.

2. One-Time Data Read
   - Reads users.txt into Member array
   - Reads reviews.txt and assigns them to the respective Member objects via composition

3. Menu navigation
   - Displays the interactive console menu to the user

4. User Selection and Processing
   - *Option 1 (Property Deep-Dive):* Searches memory for a specific hotel and calculates average and highest ratings.
   - *Option 2 (Traveler Profile):* Retrieves a specific user by ID and displays their details and authored reviews.
   - *Option 3 (Add Review):* Prompts for new review input and appends it to a member's composition array in memory.
   - *Option 4 (Top-Pick Matcher):* Aggregates all reviews in memory to rank hotels and find the highest-rated property.
   - *Option 5 (Top Reviewers):* Uses a Friend Function to access private data and calculate loyalty and review metrics.

5. System Termination and One-Time Data Write
   - On exit *(Option 6)*, the program writes the updated in-memory arrays back into users.txt and reviews.txt before shutting down.

###4.2 Class Definition
---
There are three classes in this program including user, member and review class.

| Class Name | Role / Responsibility |
| :--- | :--- |
| **User** | The **Base Class**. Acts as the foundational blueprint for a person using the system.<br>It encapsulates standard biographical data (ID, name, state, email)<br>and holds an array of `Review` objects (Composition). |
| **Member** | The **Derived Class**. Inherits from `User` and represents a registered traveler.<br>It adds specialized loyalty attributes (`memberType`, `userPoint`).<br>It internally encapsulates all business logic for processing detailed reviews<br>(`countWords`, `addReviewWithPoints`) and houses the static processing methods<br>for deep-dives and matchers. |
| **Review** | The **Composition Class**. Represents a single hotel review.<br>It is instantiated and "owned" by the `User` class.<br>It overloads relational operators to allow easy comparison of hotel ratings. |

###4.5 C++ Code

```python
%%writefile User.h
#ifndef USER_H
#define USER_H

#include <string>
#include "Review.h"
#include <iostream>
#include <iomanip>

using namespace std;

const int MAX_REVIEWS = 50;

class User {
protected: // Protected so derived classes can access if needed
    string id;
    string name;
    string state;
    string email;

    // composition
    Review reviews[MAX_REVIEWS];
    int reviewCount;

public:
    User() : id(""), name(""), state(""), email(""), reviewCount(0) {}
    User(string id, string name, string state, string email)
        : id(id), name(name), state(state), email(email), reviewCount(0) {}

    string getId() const { return id; }
    string getName() const { return name; }
    string getState() const { return state; }
    string getEmail() const { return email; }
    int getReviewCount() const { return reviewCount; }

    void setId(string id) { this->id = id; }
    void setName(string name) { this->name = name; }
    void setState(string state) { this->state = state; }
    void setEmail(string email) { this->email = email; }

    void addReview(const Review& review) {
        if (reviewCount < MAX_REVIEWS) {
            reviews[reviewCount++] = review;
        } else {
            cout << "Maximum reviews reached for user " << name << endl;
        }
    }

    Review getReview(int index) const {
        if (index >= 0 && index < reviewCount) {
            return reviews[index];
        }
        return Review(); // Return empty if out of bounds
    }

    void displayAuthoredReviews() const {
        cout << "\n--- Reviews by " << name << " ---" << endl;
        if (reviewCount == 0) {
            cout << "No reviews authored yet." << endl;
        } else {
            for (int i = 0; i < reviewCount; i++) {
                cout << i + 1 << ". ";
                reviews[i].displayReview();
                cout << "----------------------------------------" << endl;
            }
        }
    }
};

#endif
```

```python
%%writefile Member.h
#ifndef MEMBER_H
#define MEMBER_H

#include <string>
#include <iostream>
#include <iomanip>
#include "User.h"
#include "Review.h"

using namespace std;

// Inherits from User class
class Member : public User {
private:
    string memberType;
    int userPoint;

    int countWords(const string& str) const {
        int words = 0;
        bool inWord = false;
        for (char c : str) {
            if (isspace(c)) {
                inWord = false;
            } else if (!inWord) {
                inWord = true;
                words++;
            }
        }
        return words;
    }

public:
    Member() : User(), memberType(""), userPoint(0) {}
    Member(string id, string name, string state, string email, string memberType, int userPoint)
        : User(id, name, state, email), memberType(memberType), userPoint(userPoint) {}

    string getMemberType() const { return memberType; }
    int getUserPoint() const { return userPoint; }

    void setMemberType(string memberType) { this->memberType = memberType; }
    void setUserPoint(int userPoint) { this->userPoint = userPoint; }

    void displayProfile() const {
        cout << "========================================" << endl;
        cout << "Member Profile: " << name << " (" << id << ")" << endl;
        cout << "State: " << state << " | Email: " << email << endl;
        cout << "Member Type: " << memberType << " | Points: " << userPoint << endl;
        cout << "Total Authored Reviews: " << reviewCount << endl;
        cout << "========================================" << endl;
    }

    // Operator Overloading - Compare loyalty points
    bool operator>(const Member& other) const {
        return this->userPoint > other.userPoint;
    }

    // Static member functions to process arrays
    static void propertyDeepDive(Member members[], int count, string targetHotel) {
        cout << "\n=== Property Deep-Dive: " << targetHotel << " ===" << endl;
        int matchCount = 0;
        int totalRating = 0;
        Review bestReview;
        bool hasBest = false;

        for (int i = 0; i < count; i++) {
            for (int j = 0; j < members[i].getReviewCount(); j++) {
                Review r = members[i].getReview(j);
                if (r.getHotelName() == targetHotel) {
                    cout << "Review by " << members[i].getName() << " (" << members[i].getMemberType() << "):" << endl;
                    cout << "Rating: " << r.getRating() << "/5" << endl;
                    cout << "Review: " << r.getReview() << endl;
                    cout << "----------------------------------------" << endl;
                    totalRating += r.getRating();
                    matchCount++;

                    // Utilize overloaded operator>
                    if (!hasBest || r > bestReview) {
                        bestReview = r;
                        hasBest = true;
                    }
                }
            }
        }

        if (matchCount > 0) {
            double average = static_cast<double>(totalRating) / matchCount;
            cout << "Summary for " << targetHotel << ":" << endl;
            cout << "Total Reviews: " << matchCount << endl;
            cout << "Average Rating: " << fixed << setprecision(1) << average << " / 5.0" << endl;
            cout << "Highest Individual Rating: " << bestReview.getRating() << "/5 (\"" << bestReview.getReview() << "\")" << endl;
        } else {
            cout << "No reviews found for hotel: " << targetHotel << endl;
        }
        cout << "========================================" << endl;
    }

    static void topPickMatcher(Member members[], int count) {
        // Collect unique hotels and their ratings
        const int MAX_HOTELS = 100;
        string hotelNames[MAX_HOTELS];
        int hotelTotalRatings[MAX_HOTELS] = {0};
        int hotelReviewCounts[MAX_HOTELS] = {0};
        int uniqueHotelCount = 0;

        for (int i = 0; i < count; i++) {
            for (int j = 0; j < members[i].getReviewCount(); j++) {
                Review r = members[i].getReview(j);
                string hName = r.getHotelName();

                // Check if hotel exists in array
                int hIndex = -1;
                for (int k = 0; k < uniqueHotelCount; k++) {
                    if (hotelNames[k] == hName) {
                        hIndex = k;
                        break;
                    }
                }

                if (hIndex == -1 && uniqueHotelCount < MAX_HOTELS) {
                    // New hotel found
                    hotelNames[uniqueHotelCount] = hName;
                    hotelTotalRatings[uniqueHotelCount] = r.getRating();
                    hotelReviewCounts[uniqueHotelCount] = 1;
                    uniqueHotelCount++;
                } else if (hIndex != -1) {
                    // Existing hotel
                    hotelTotalRatings[hIndex] += r.getRating();
                    hotelReviewCounts[hIndex]++;
                }
            }
        }

        if (uniqueHotelCount == 0) {
            cout << "No reviews available to match top picks." << endl;
            return;
        }

        // Find the highest rated hotel
        double maxAverage = -1.0;
        string topHotel = "";
        int topReviews = 0;

        cout << "\n=== Top-Pick Matcher: Hotel Rankings ===" << endl;
        for (int i = 0; i < uniqueHotelCount; i++) {
            double avg = static_cast<double>(hotelTotalRatings[i]) / hotelReviewCounts[i];
            cout << "Hotel: " << hotelNames[i] << " | Avg Rating: " << fixed << setprecision(1) << avg << " (" << hotelReviewCounts[i] << " reviews)" << endl;

            if (avg > maxAverage) {
                maxAverage = avg;
                topHotel = hotelNames[i];
                topReviews = hotelReviewCounts[i];
            }
        }

        cout << "\n*** Highest-Rated Hotel ***" << endl;
        cout << topHotel << " with an average rating of " << maxAverage << " (" << topReviews << " reviews)!" << endl;
        cout << "========================================" << endl;
    }

    static Member* findMemberByID(Member members[], int count, string id) {
        for (int i = 0; i < count; i++) {
            if (members[i].getId() == id) {
                return &members[i];
            }
        }
        return nullptr;
    }


    void addReviewWithPoints(const Review& review) {
        addReview(review);

        int pointsEarned = 100; // Base points
        int wordCount = countWords(review.getReview());

        if (wordCount > 50) {
            pointsEarned += 10; // Bonus for detailed review
            cout << "\nAwesome! You wrote " << wordCount << " words and earned a 10-point detailed review bonus!" << endl;
        }

        // Update points
        this->userPoint += pointsEarned;

        cout << "Review added successfully! You earned " << pointsEarned << " points." << endl;
        cout << "New Total Points: " << this->userPoint << endl;
    }

    // Friend Functions: allow an external utility to access private class members
    friend void topReviewersAnalytics(Member members[], int count);
};

#endif
```

```python
%%writefile Review.h
#ifndef REVIEW_H
#define REVIEW_H

#include <string>
#include <iostream>

using namespace std;

class Review {
private:
    string id;
    int rating;
    string review;
    string hotelName;

public:
    Review() : id(""), rating(0), review(""), hotelName("") {}
    Review(string id, int rating, string review, string hotelName) : id(id), rating(rating), review(review), hotelName(hotelName) {}

    string getId() const { return id; }
    int getRating() const { return rating; }
    string getReview() const { return review; }
    string getHotelName() const { return hotelName; }

    void setId(string id) { this->id = id; }
    void setRating(int rating) { this->rating = rating; }
    void setReview(string review) { this->review = review; }
    void setHotelName(string hotelName) { this->hotelName = hotelName; }

    // Operator Overloading: Compare ratings of two reviews
    bool operator>(const Review& other) const {
        return this->rating > other.rating;
    }
    bool operator<(const Review& other) const {
        return this->rating < other.rating;
    }

    void displayReview() const {
        cout << "User ID: " << id << " | Hotel: " << hotelName << " | Rating: " << rating << "/5" << endl;
        cout << "Review: " << review << endl;
    }
};

#endif
```

```python
%%writefile main.cpp
#include <iostream>
#include <fstream>
#include <string>
#include "User.h"
#include "Member.h"
#include "Review.h"

using namespace std;

// Friend Function Implementation
void topReviewersAnalytics(Member members[], int count) {
    cout << "\n=== Top Reviewers Analytics ===" << endl;
    if (count == 0) {
        cout << "No members found." << endl;
        return;
    }

    int maxReviews = -1;
    for (int i = 0; i < count; i++) {
        if (members[i].getReviewCount() > maxReviews) {
            maxReviews = members[i].getReviewCount();
        }
    }

    cout << "Members with the highest number of reviews (" << maxReviews << "):" << endl;
    for (int i = 0; i < count; i++) {
        if (members[i].getReviewCount() == maxReviews) {
            cout << "- " << members[i].getName() << " (" << members[i].getId()
                 << ") - Member Type: " << members[i].getMemberType()
                 << " | Points: " << members[i].userPoint << endl;
        }
    }

    Member mostLoyal = members[0];
    for (int i = 1; i < count; i++) {
        if (members[i] > mostLoyal) {
            mostLoyal = members[i];
        }
    }
    cout << "\n* Most Loyal Member (Highest Points): " << mostLoyal.getName()
         << " with " << mostLoyal.userPoint << " points!" << endl;

    cout << "===============================" << endl;
}

int main() {
    const int MAX_MEMBERS = 10;
    Member members[MAX_MEMBERS];
    int memberCount = 0;

    // read users.txt
    ifstream userFile("users.txt");
    if (userFile) {
        string id, name, state, email, memberType, pointsStr;

        while (memberCount < MAX_MEMBERS && getline(userFile, id, '\t')) {
            getline(userFile, name, '\t');
            getline(userFile, state, '\t');
            getline(userFile, email, '\t');
            getline(userFile, memberType, '\t');
            getline(userFile, pointsStr);

            int points = 0;
            try {
                if (!pointsStr.empty())
                points = stoi(pointsStr);
            } catch (...) {}

            if (!id.empty()) {
                members[memberCount++] = Member(id, name, state, email, memberType, points);
            }
        }
        userFile.close();
    } else {
        cout << "Warning: Could not open users.txt" << endl;
    }

    // read reviews.txt
    ifstream reviewFile("reviews.txt");
    if (reviewFile) {
        string id, ratingStr, reviewText, hotelName;

        while (getline(reviewFile, id, '\t')) {
            getline(reviewFile, ratingStr, '\t');
            getline(reviewFile, reviewText, '\t');
            getline(reviewFile, hotelName);

            int rating = 0;
            try {
                if (!ratingStr.empty())
                  rating = stoi(ratingStr);
            } catch (...) {}

            Member* m = Member::findMemberByID(members, memberCount, id);
            if (m != nullptr) {
                m->addReview(Review(id, rating, reviewText, hotelName));
            }
        }
        reviewFile.close();
    } else {
        cout << "Warning: Could not open reviews.txt" << endl;
    }

// menu
    int choice = 0;
    while (choice != 6) {
        cout << "\n--- Hotel Review System ---" << endl;
        cout << "1. Property Deep-Dive" << endl;
        cout << "2. Traveler Profile" << endl;
        cout << "3. Add a New Review" << endl;
        cout << "4. Top-Pick Matcher" << endl;
        cout << "5. Top Reviewers Analytics" << endl;
        cout << "6. Exit" << endl;
        cout << "Enter choice: ";

        if (!(cin >> choice)) {
            cin.clear();
            cin.ignore(10000, '\n');
            cout << "Invalid input. Please enter a number." << endl;
            continue;
        }
        cin.ignore(10000, '\n');

        if (choice == 1) {
            cout << "Enter Hotel Name: ";
            string hotel;
            getline(cin, hotel);
            Member::propertyDeepDive(members, memberCount, hotel);
        } else if (choice == 2) {
            cout << "Enter Traveler ID: ";
            string tid;
            getline(cin, tid);
            Member* m = Member::findMemberByID(members, memberCount, tid);
            if (m) {
                m->displayProfile();
                m->displayAuthoredReviews();
            } else {
                cout << "Traveler not found." << endl;
            }
        } else if (choice == 3) {
            cout << "Enter Traveler ID: ";
            string tid;
            getline(cin, tid);
            Member* m = Member::findMemberByID(members, memberCount, tid);

            if (m) {
                string newHotel, newReviewText;
                int newRating;

                cout << "Enter Hotel Name: ";
                getline(cin, newHotel);

                cout << "Enter Rating (1-5): ";
                while (!(cin >> newRating) || newRating < 1 || newRating > 5) {
                    cin.clear();
                    cin.ignore(10000, '\n');
                    cout << "Invalid input! Please enter a number between 1 and 5: ";
                }
                cin.ignore(10000, '\n');

                cout << "Enter Review: ";
                getline(cin, newReviewText);

                m->addReviewWithPoints(Review(tid, newRating, newReviewText, newHotel));

            } else {
                cout << "Traveler not found. Cannot add review." << endl;
            }
        } else if (choice == 4) {
            Member::topPickMatcher(members, memberCount);
        } else if (choice == 5) {
            topReviewersAnalytics(members, memberCount);
        } else if (choice == 6) {
            cout << "Exiting..." << endl;
        } else {
            cout << "Invalid choice. Please try again." << endl;
        }
    }

    cout << "Saving data..." << endl;

        // write users.txt
        ofstream outUserFile("users.txt");
        if (outUserFile) {
            for (int i = 0; i < memberCount; i++) {
                outUserFile << members[i].getId() << "\t"
                            << members[i].getName() << "\t"
                            << members[i].getState() << "\t"
                            << members[i].getEmail() << "\t"
                            << members[i].getMemberType() << "\t"
                            << members[i].getUserPoint() << "\n";
            }
            outUserFile.close();
        } else {
            cout << "Error: Could not save users.txt" << endl;
        }

        // write reviews.txt
        ofstream outReviewFile("reviews.txt");
        if (outReviewFile) {
            for (int i = 0; i < memberCount; i++) {
                for (int j = 0; j < members[i].getReviewCount(); j++) {
                    Review r = members[i].getReview(j);
                    outReviewFile << r.getId() << "\t"
                                  << r.getRating() << "\t"
                                  << r.getReview() << "\t"
                                  << r.getHotelName() << "\n";
                }
            }
            outReviewFile.close();
        } else {
            cout << "Error: Could not save reviews.txt" << endl;
        }

        cout << "Save complete." << endl;

        return 0;
}
```
